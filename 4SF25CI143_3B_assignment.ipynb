```python
data = np.random.randint(1, 15, (4, 3))

df = pd.DataFrame(data, columns=['X', 'Y', 'Z'])

print("Original Data:")
print(df)

print("Row-wise Sum:")
print(df.sum(axis=1))

print("Column-wise Sum:")
print(df.sum(axis=0))
```

```python
data = np.random.randint(1, 20, (5, 3))

df = pd.DataFrame(data, columns=['A', 'B', 'C'])

df.iloc[0, 1] = np.nan
df.iloc[2, 2] = np.nan
df.iloc[4, 0] = np.nan

print("Before Filling:")
print(df)

df = df.fillna(df.mean())

print("After Filling:")
print(df)
```

```python
data = np.random.randint(1, 30, (6, 4))

df = pd.DataFrame(data, columns=['A', 'B', 'C', 'D'])

df.iloc[1, 2] = np.nan
df.iloc[3, 0] = np.nan

print("Original DataFrame:")
print(df)

new_df = df.dropna()

print("After Removing NaN Rows:")
print(new_df)
```

```python
df = pd.DataFrame({
    'Category': np.random.choice(['A', 'B', 'C'], 10),
    'Value': np.random.randint(20, 100, 10)
})

print("Original Data:")
print(df)

result = df.groupby('Category')['Value'].agg(['sum', 'mean'])

print("Category-wise Result:")
print(result)
```

```python
df = pd.DataFrame({
    'Product': ['Pen', 'Book', 'Bag', 'Pencil', 'Bottle', 'Box'],
    'Category': ['Stationery', 'Stationery', 'Accessories',
                 'Stationery', 'Accessories', 'Accessories'],
    'Sales': np.random.randint(100, 1000, 6)
})

print("Sales Data:")
print(df)

result = df.groupby('Category')['Sales'].sum()

print("Total Sales:")
print(result)
```

```python
data1 = pd.DataFrame({
    'RollNo': [101, 102, 103, 104],
    'Name': ['Arun', 'Ravi', 'Kiran', 'Manu']
})

data2 = pd.DataFrame({
    'RollNo': [101, 102, 103, 105],
    'Marks': [85, 90, 75, 88]
})

merged = pd.merge(data1, data2, on='RollNo')

print("Merged DataFrame:")
print(merged)
```

```python
index = pd.MultiIndex.from_tuples([
    ('Class1', 'Student1'),
    ('Class1', 'Student2'),
    ('Class2', 'Student1'),
    ('Class2', 'Student2')
], names=['Class', 'Student'])

df = pd.DataFrame({
    'Marks': [80, 90, 75, 85],
    'Age': [18, 19, 18, 20]
}, index=index)

print("Complete DataFrame:")
print(df)

print("Specific Student:")
print(df.loc[('Class1', 'Student1')])

print("Class1 Details:")
print(df.loc['Class1'])
```

```python
index = pd.MultiIndex.from_product(
    [['A', 'B'], ['X', 'Y']],
    names=['Category', 'SubCategory']
)

df = pd.DataFrame({
    'Value': np.random.randint(10, 50, 4)
}, index=index)

print("Original DataFrame:")
print(df)

result = df.groupby(level=['Category', 'SubCategory']).sum()

print("Grouped Sum:")
print(result)
```

```python
data = np.random.randint(1, 15, (5, 3))

df = pd.DataFrame(data, columns=['P', 'Q', 'R'])

print("Original Data:")
print(df)

result = df.apply(lambda x: x * 2)

print("After Doubling:")
print(result)
```

```python
words = pd.Series([
    'mango',
    'orange',
    'apple',
    'grapes',
    'banana'
])

print("Original Strings:")
print(words)

upper_words = words.str.upper()

print("Uppercase Strings:")
print(upper_words)
```
